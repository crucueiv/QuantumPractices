````python
#
# RUTINAS matplotlib para dibujar ejes, vectores y números complejos como fasor
# Basadas en el material disponible en 
# https://talentq-es.github.io/TalentQ-CICC/docs/index.html
#
def plot_2D_plane(right=1,up=1,left=-1,down=-1,fsize=(8,8)):
    hpoints, vpoints = [],[]
    for i in range(left,right+1):
        if i!=0: hpoints.append(i)
    for i in range(down,up+1):
        if i!=0: vpoints.append(i)
    
    ax = plt.figure(figsize=fsize).gca()

    # Establece escalas iguales para ambos ejes
    ax.set(xlim=(left-0.5,right+0.5), ylim=(down-0.5, up+0.5), aspect='equal')
    # Elimina los bordes superior e inferior
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    # Establece los bordes inferior e izquierdo como ejes de coordenadas
    ax.spines['bottom'].set_position('zero')
    ax.spines['left'].set_position('zero')
    # Crea ticks en cada entero para habilitar dibujo de un retículo menor
    # Create minor ticks placed at each integer to enable drawing of minor grid
    ax.set_xticks(hpoints)
    ax.set_yticks(vpoints)
    # Dibuja retículo
    ax.grid(which='both', color='green', linewidth=1, linestyle='-', alpha=0.2)
    # Crea etiquetas 'x' e 'y' al final de cada eje
    ax.set_xlabel('x', size=14, labelpad=-24, x=1.03)
    ax.set_ylabel('y', size=14, labelpad=-21, y=1.02, rotation=0)
    # Dibuja flechas
    arrow_fmt = dict(markersize=4, color='black', clip_on=False)
    ax.plot((1), (0), marker='>', transform=ax.get_yaxis_transform(), **arrow_fmt)
    ax.plot((0), (0), marker='<', transform=ax.get_yaxis_transform(), **arrow_fmt)
    ax.plot((0), (1), marker='^', transform=ax.get_xaxis_transform(), **arrow_fmt)
    ax.plot((0), (0), marker='v', transform=ax.get_xaxis_transform(), **arrow_fmt)


def draw_sides(x=1,y=2,side_color="b",lwidth=1):
    plt.arrow(x,0,0,y,color=side_color,linestyle="dotted",width=0.001*lwidth)
    plt.arrow(0,y,x,0,color=side_color,linestyle="dotted",width=0.001*lwidth)
        
def draw_vector(x=1,y=0,vname="v",show_name=False,vcolor="b",sides=False,side_color="b",lwidth=1,zorigin=0):
    x = x
    y = y
    plt.quiver(0+zorigin.real,0+zorigin.imag,x,y,scale=1,scale_units='xy',angles = 'xy',color=vcolor,width=0.008*lwidth)
    dx = x
    if y<0: dy=y-0.3
    else: dy = y+0.3
        
    if show_name:
        vector_name="$"+vname+"=("+str(x)+","+str(y)+")$"
        plt.text(dx,dy,vector_name,color=vcolor)
    
    if sides:
        draw_sides(x,y,side_color)
        
def plot_complex_number(z,show_name=False,vcolor="b",sides=False,side_color="b",lwidth=1.4, size_cm=10, zorigin=0, hold=False):
     x = z.real
     y = z.imag
     size_in = size_cm / 2.54
     if not hold:
         plot_2D_plane(left=-int(abs(x))-1,right=int(abs(x))+1,up=int(abs(y))+1,down=-int(abs(y))-1,fsize=(size_in,size_in))
     draw_vector(x=x,y=y,vname=False,show_name=False,vcolor=vcolor,sides=False,side_color=side_color,lwidth=lwidth,zorigin=zorigin)
    
def plot_cplx(z, color="b", size_cm=10, zorigin=0, hold=False):
    """
        z = complex number
        color = a color string
        size_cm = figure size in cm
        zorigin = complex where the vector starts 
        hold = if hold=True the complex is drawn in the same axis as the previous one 
    """
    plot_complex_number(z, vcolor=color, size_cm=size_cm, zorigin=zorigin, hold=hold);
````

### Conversión entre formas cartesiana y polar

La conocidad como fórmula de Euler permite convertir de representación polar a cartesiana:

$$
z = r e^{i\theta} = x + i y ~~~\hbox{ con }  ~~~\left\{\begin{array}{l} x=r \cos \theta \\ \rule{0mm}{4mm} y = r\sin \theta
\end{array} \right.
$$

La conversión inversa, de cartesiana a polar sería la siguiente:

$$
z = x + i y  = r e^{i\theta} ~~~ \hbox{ con } ~~~ \left\{\begin{array}{l} r=\sqrt{x^2+y^2} \\  \rule{0mm}{4mm} \theta = {\rm arctan2}(y,x)
\end{array} \right.
$$

Donde $\text{arctan2}(y,x)$ es básicamente la función arcotangente de $\frac{y}{x}$ pero añade un ángulo de ajuste en función del cuadrante donde se ubican el número complejo, que viene determinado por los signos de $x$ e $y$. También tiene en cuenta casos asintóticos como $x=0$.

        """Convierte de cartesianas a polar

Uso: 

Coordenadas cartesianas de $z= 4 e^{5/4\pi j}$: ```x,y = p2c(4, np.pi*5/4)```

Coordenadas polares de $3-5j$: ```r,th = c2p(3, -5)```

````python
##### SOLUCIÓN

def p2c(r,theta):
    """Convierte de polar a cartesiana
     debe devolver dos valores x e y
    """
    x = r * np.cos(theta)
    y = r * np.sin(theta)
    return x, y 

    
def c2p(x,y):
    """Convierte de cartesiana a polar
     debe devolver dos valores r,theta
    """
    r = np.sqrt(x*x + y*y)
    th = arcotangentediv(y,x) 
    return r, th

def arcotangentediv(y,x):
    # Basicamente es el arctan(y/x) pero corrigiendo el signos en función del cuadrante y los casos
    # asintóticos x=0
    if x==0:
        if y>0:  #si x=0 caso asintótico, depende del signo de y 
            th = np.pi/2
        elif y<0:
            th = 3*np.pi/2  
    elif x>0 and y>=0:   #primer cuadrante
        th = np.atan(y/x)
    elif x<0 and y>=0:   #segundo cuadrante
        th = np.atan(y/x) + np.pi
    elif x<0 and y<0:    #tercer cuadrante
        th = np.atan(y/x) + np.pi
    elif x>0 and y<0:    #cuarto cuadrante
        th = np.atan(y/x) + 2*pi

    # Normallizar [-pi, pi]
    if th > np.pi:
        th = th - 2*np.pi
    return th

# ¡Atención! la función arcotangente2(y,x) ya está implementada en numpy: np.a2(y,x)
# Comprueba que sale este resultado
````

````python
# Generemos un complejo a partir de su representación polar, módulo y ángulo
pi = np.pi
R = 2
T = 2*pi*3/7
xx,yy = p2c(R, T)
zz = xx + 1j*yy
print(f'zz={zz}')
plot_cplx(zz, size_cm=8)
plt.show()

# Regeneremos el número, obteniendo su representación polar a partir de sus componentes real e imaginaria
rr, tt = c2p(zz.real, zz.imag)
print(f"rho={rr:.3f} theta={tt:.3f}")
zz2 = rr * np.exp(1j * tt)        # número a partir de módulo y fase 
plot_cplx(zz2, "violet", size_cm=8)
plt.show()
````

### Conjugacion compleja
 
Todo número complejo, $z$, lleva *asociado* otro, $z^*$, denominado el *complejo conjugado* que se obtiene cambiando $\i \to -\i$
<br>

$$
z = x+\i y ~~~~\leftrightarrow~~~~ z^* = x - \i y \hspace{2cm}
$$

Es evidente que $(z^*)^* = z$. Es decir, la conjugación compleja es una *involución*.

````python
# Conjugación compleja

# Número complejo aleatorio con distribución normal de desviación unidad
z = np.random.normal(0,1) + 1j*np.random.normal(0,1)
z = np.round(z, 3)  # redondeo a 3 decimales
plot_cplx(z,"red")

# Su conjugado
zc = z.conjugate()

print('z=', z, ' | z* = ',zc)
plot_cplx(zc, "turquoise", hold=True)
````

Observa que en forma polar la *conjugación compleja* se obtiene *cambiando el signo* de la fase
<br>

$$
z = \rho e^{i\theta} =  \rho \cos\theta + \i \rho \sin\theta ~~\longleftrightarrow ~~ z^* = \rho e^{-i\theta} = \rho \cos\theta - \i \rho \sin\theta \hspace{4cm}
$$

## Operaciones básicas  

Los numeros complejos ${\mathbb C}$ forman una estructura matemática, denominada *cuerpo*, que admiten dos operaciones *internas*: la *suma* y la *multiplicación*. Vamos a estudiarlas por separado

### Suma

En representación *cartesiana* se *suman las partes real e imaginaria por separado*

$$
(a + \i b) + (c + \i d) = (a+c) + \i (b+d)
$$ 


La resta es obvia, ya que $a,b,c,d$ pueden ser números negativos. 


En *forma polar*, la suma de dos números complejos no admite ninguna simplificación, y deben transformarse primeramente a forma cartesiana, para sumarse. 

$$
z + w = \rho e^{i\theta} + \sigma e^{i\phi} = (\rho\cos\theta + \sigma\cos\phi) + i(\rho\sin\theta +  \sigma\sin\phi) \hspace{6cm}
$$

###  Multiplicación 



En *forma cartesiana* la multiplicación es complicada, debiendo multiplicarse todos los factores entre sí, y
teniendo en cuenta que $\i^2= -1$

$$
(a + \i b) (c + \i d) =ac +  a\i d +\i bc +\i^2 bd = (ab - bd) + \i(ac + bd)\hspace{6cm}
$$ 

<br>
    
En *forma polar* la multiplicación es simple: se multiplican los módulos y se suman las fases

$$
 r e^{i\theta} s e^{i\phi} = rs\,   e^{i(\theta + \phi)} 
$$

### División


Al igual que la multiplicación, en forma cartesiana, la división *no es simple*. Sea $z = a+ \i b$ y $w=c+\i d$ 
<br>
<br>

$$
\frac{z}{w} = \frac{z}{w}\frac{w^*}{w^*} = \frac{( a+ \i b)(c-\i d)}{|w|^2} = \frac{ac+bd + \i(bc-ad)}{c^2+d^2}  = \frac{ac+bd}{c^2+d^2} +\i\frac{bc-ad}{c^2+d^2}
$$

En forma polar la división es tan sencilla como la multiplicación. Se toma el  cociente de los módulos y la resta de las fases

$$
\frac{z}{w} = \frac{\rho e^{i\theta}}{\sigma e^{i\phi}} = \frac{\rho}{\sigma}
e^{i(\theta-\phi)}
$$


````python
# Suma y resta

# Dos números complejos 
z1 = 1+3j
z2 = -2+2j

# Sumar y restar
zs = z1+z2
zd = z1-z2

# Pretty (latex) print
display(Markdown('---'))
display(Latex(r"$z_s=z_1+z_2={}+{}={}$".format(z1,z2,zs)))
display(Latex(r"$z_d=z_1-z_2={}+{}={}$".format(z1,z2,zd)))
display(Markdown('---'))

mx = max(np.abs(z1), np.abs(z2), np.abs(zs), np.abs(zd))
plot_cplx(mx-1j*mx, "none", size_cm=10) # Dirty trick: set axis as big as the biggest
plot_cplx(z1, "blue",   hold=True)
plot_cplx(z2, "orange", hold=True)
plot_cplx(z2, "orange", hold=True, zorigin=z1)
plot_cplx(zs, "green",  hold=True)
plot_cplx(zd, "gray",   hold=True)

# Observa la gráfica e interpreta geométricamente la suma y resta de complejos
````

### Valor absoluto

Se define el *valor absoluto* de un número complejo $z=x + jy$:

$$
|z| = \sqrt{z z^*} = \sqrt{x^2 + y^2}
$$

El valor absoluto de un número complejo  escrito en forma polar coincide con su  <i>módulo</i>

$$
|z| = \sqrt{zz^*} = \sqrt{\rho e^{i\theta} \rho e^{-i\theta}}=\sqrt{\rho^2} = \rho
$$
    
En consecuencia, el valor absoluto de una <i>fase pura</i> $z = e^{i\theta}$ es $\rho =1$

## Fases equiespaciadas

Un caso de interés es el de la suma de una secuencia de fases puras equiespaciadas cubriendo el intervalo $[0,2\pi[$, que ilustra el fenómeno de la _interferencia_.

### Sumas nulas

La _interferencia cuántica_, en ultima instancia, procede de la posibilidad de sumar muchas fases complejas de manera que se cancelen mutuamente.
Sea $N$ un número entero positivo, se verifica que

$$
\sum_{k=0}^{N-1} e^{2\pi \i k/N} =   e^{2\pi \i\, 0/N} +  e^{2\pi \i\, 1/N}  +~...~ + ~  e^{2\pi \i\, (N-1)/N} ~=~  ~0
 $$

Vamos a representar los números complejos y calcular su suma. Puedes cambiar $N$ y también multiplicar por un módulo constante $\rho$ y observar el resultado.
 

````python
# modifica el número N
N=7   # Número de elementos en la serie
rho=1 # Módulo: dejar a 1, o sea, suma de fases puras

π = np.pi 

#Creamos la secuencia de fases equiespaciadas
listN = np.array(range(0,N))          # Lista de números correlativos 0,1,2... N-1
listN2pi = listN * (2*π/N)            # Factor de escala 2π/N: 0, 2π/N, 4π/N, 6π/N ... (N-1)π/N  
phase_list = rho*np.exp(1j*listN2pi)  # Exponencial de cada elemento de la lista e^0, e^i2π/N...
#print('lista de fases =', np.round(phase_list,2)) # debug

# Impresión fantasma para poner un tamaño adecuado inicial a los ejes
plot_cplx(rho*(1+1j)*(1/2 if rho<=1 else 1), "none", size_cm=10)
  
# Dibujamos la serie de números complejos generada
for z in phase_list:
    plot_cplx(z, hold=True)

# Dibujar circulo de radio rho 
plt.gca().add_patch(plt.Circle((0.,0.),rho,color='black',fill=False)) 
   
# Suma de la serie de fases puras 
S = np.round(sum(rho*phase_list),10) 
display(Latex(r"$\displaystyle \sum_{k=0}^{%i} e^{\large\frac{2\pi j k}{N}} = %.2f + %.2fj$"%(N-1, S.real, S.imag)))
````

Aun podemos obtener sumas nulas (interferencias)  más generales.
Sea  $1\leq m \leq N-1$ un *número entero* por el que multiplicamos *todas las fases*. El resultado es el mismo

$$
\sum_{k=0}^{N-1} e^{2\pi \i \, m k/N} =   e^{2\pi \i \, 0 m/N} +  e^{2\pi \i \, 1 m/N} + e^{2\pi   \i \, 2  m/N}  +~...~ +   e^{2\pi \i\, (N-1) m/N} ~=~  ~0
 $$


Sin embargo si elegimos $m$ como múltiplo de $N$, $m = 0, N, 2N,... $, entonces la suma no se anula y su valor es igual a $N$. 

Tomemos por ememplo $m=N$ 

$$
\sum_{k=0}^{N-1} e^{2\pi \i \, k N/N} = \sum_{k=0}^{N-1} e^{2\pi \i  k}  =  \sum_{k=0}^{N-1} 1 =~  ~N
 $$


````python
# modifica el número N
N=7
rho=1
π = np.pi 

# Nuevo parámetro m: probar poniendo múltiplos de N y no múltiplos de N, para observar el efecto
m=N

#Creamos la secuencia de fases equiespaciadas
listN = np.array(range(0,N))         # Lista de números correlativos 0,1,2... N-1
listN2pi = listN * (2*π/N) * m       # Factor de escala 2π/N*m: 0, 2π/N*m, 4π/N*m, 6π/N*m ... (N-1)π/N*m  
phase_list = rho*np.exp(1j*listN2pi) # Exponencial de cada elemento de la lista e^0, e^i*m*2π/N...
#print('lista de fases =', np.round(phase_list,2)) # debug

# Impresión fantasma para poner un tamaño adecuado inicial a los ejes
plot_cplx(rho*(1+1j)*(1/2 if rho<=1 else 1), "none", size_cm=10)
  
# Dibujamos la serie de números complejos generada
for z in rho*phase_list:
    plot_cplx(z, hold=True)

# Dibujar circulo de radio rho 
plt.gca().add_patch(plt.Circle((0.,0.),rho,color='black',fill=False)) 
   
# Suma de la serie de fases puras 
S = np.round(sum(rho*phase_list),10) 
display(Latex(r"$\displaystyle \sum_{k=0}^{%i} e^{\large\frac{2\pi m k}{N}} = %.2f + %.2fm$"%(N-1, S.real, S.imag)))
````

## Función de onda simple

````python
# En este ejemplo se ilustra como una función exponencial compleja
# representa una función de onda

display(Markdown(r'### Demo de y(t) = $\large A e^{i\,wt}$ como onda'))

# Eje X  e Y a dibujar 
t = np.linspace(-2, 2, 1000)   # Array con las muestras del eje de tiempo
A = 1                  # Amplitud
T=1                    # Periodo
f = 1/T                # Frecuencia
w = 2*np.pi*f          # Frecuencia angular
y = np.exp(1j * w * t) # Funcion y(t)= A e^(i wt) 

# Dibujemos la onda: esto es la parte real de y(t)
fig, axes = plt.subplots(1, 1, figsize=(6,3))
ax = axes         # Como el subplot es 1x1 la variable ax es un escalar,
                  # si fuera 1x2 sería una lista y nxn una matriz
ax.set_title(r"Función de onda $y(t)=\Re(e^{i wt})$")
ax.plot(t, y.real, color="blue", linewidth=2)  # Funcion de onda: parte real de Ae^(j wt)
ax.plot(t, 0*t, "k:", linewidth=0.5)           # Eje x
ax.plot([0,0], [-A,A], "k:", linewidth=0.5)    # Eje y
plt.xlabel('t')

plt.show()

# Generemos diferentes fases en el intervalo [0, pi] equiespaciadas np.pi/6
theta = np.arange(0, np.pi+1e-10, np.pi/6)
# Una lista de estilos de trazo, a elegir
styles = [color+trazo for trazo in ['', ':', '--'] for color in ['k','b','r','g','y'] ]
print("Fases=", theta)

fig, axes = plt.subplots(1, 1, figsize=(6,3))
ax = axes
ax.set_title(r"Función de onda desfasada $\theta$: $y(t)=\Re(e^{i wt} e^{i\theta})=\Re(e^{i wt + \theta})$")
c = 0
for th in theta:
    ytheta = y * np.exp(1j * th) # Multiplicar por e^(i theta), o sea añadir esta fase
    thpi = np.round(th/np.pi,2)  # a efectos de presentación, veces que th es pi
    ax.plot(t, ytheta.real, styles[c%len(styles)], label=rf'$\theta$ ={thpi}$\pi$', linewidth=1)
    c = c+1
plt.legend(bbox_to_anchor=(1, 1),loc='upper left')
plt.xlabel('t')
plt.grid()
plt.show()
````
